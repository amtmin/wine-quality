### Базовая статистика по числовым столбцам

`.describe()` показывает для каждого числового столбца:
- count - сколько непустых значений
- mean - среднее
- std - стандартное отклонение (разброс)
- min, 25%, 50%, 75%, max - квантили 
> Квантили - это значения, которые делят отсортированный набор данных на четыре равные части.

Это быстрый способ увидеть, «как выглядят» числа:
есть ли странные значения, большой ли разброс и т.д.

```python
df.describe()
```

### Выводы по обзору данных
